```python
import numpy as np
import matplotlib.pyplot as plt

# Number of grid points
Nx = 51
Nz = 51

# Domain
x = np.linspace(0, 1, Nx)
z = np.linspace(0, 1, Nz)

# Create 2-D grid
X, Z = np.meshgrid(x, z)# specifically used to create a 2D coordinate grid from 1D coordinate array

# Print information
print("Number of x points:", Nx)
print("Number of z points:", Nz)
print("Grid shape:", X.shape)

# Plot grid
plt.figure(figsize=(7, 5))

plt.scatter(X, Z, s=5)

plt.xlabel("x")
plt.ylabel("z")
plt.title("Computational Grid")

plt.grid()

plt.savefig("computational_grid.png",
            dpi=300,6
            bbox_inches="tight")
plt.show()
```

```python
#what np.meshgrid do
# instead of having just x and z we will have X(x,z) and Z(x,z) 
```

```python
dx = x[1] - x[0]
dz = z[1] - z[0]

print("dx =", dx)
print("dz =", dz)
```

```python
#Creating the temperature field(T)
T=1-Z
print(T)
#this represents the analytical conduction solution (T(Z)=1-Z
```

```python
#plotting the temperature field
plt.figure(figsize=(5, 5))

contour = plt.contourf(X, Z, T, levels=20)

plt.colorbar(contour, label="Temperature")

plt.xlabel("x")
plt.ylabel("z")

plt.title("Conduction Temperature Field")
plt.savefig("conduction_temperature.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
#From the figure we can see the temperature varies only with z and not with x 
```

```python
#plotting the vertical temperature profile
plt.figure(figsize=(6, 5))

mid = Nx // 2

plt.plot(T[:, mid], z)

plt.xlabel("Temperature")
plt.ylabel("z")

plt.title("Vertical Temperature Profile")

plt.grid()

plt.savefig("temperature_profile.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
#it demonstrate the analytical conduction solution
```

```python
#Checking the boundary conditon
print("Bottom temperature:", T[0, 0])
print("Top temperature:", T[-1, 0])
```

```python
#Calculating the temperature gradient ie,dT/dz = -1
dTdz = np.gradient(T, dz,axis=0)
print("Temperature gradient at center:")
print(dTdz[Nx//2,Nz//2])
#will be useful for later because heat flux is related to temp gradient through Fourie's law
#q= -k dT/dz
```
